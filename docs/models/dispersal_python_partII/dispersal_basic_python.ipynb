## Curent model description
This is a spatial dispersal model where agents represent organisms or particles that reproduce and spread across a landscape. Agents start clustered in a small corner area (6x6 cells) and each simulation step, they probabilistically "hatch" new offspring into neighboring empty cells based on their growth rate. Over time, the population expands outward from the initial cluster, creating a spreading pattern across the grid. The model allows to set number of agents, size of the grid, and growth rate.
